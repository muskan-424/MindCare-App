# Gemini 2.5 Flash Token & Limits Test

This notebook tests the Gemini API to check token usage per request.

**Free Tier Limits for Gemini 2.5 Flash:**
- 15 Requests Per Minute (RPM)
- 1,000,000 Tokens Per Minute (TPM)
- 1,500 Requests Per Day (RPD)

```python
import requests
import json

# IMPORTANT: Paste your Google API Key here
API_KEY = "PASTE_YOUR_API_KEY_HERE"
MODEL = "gemini-2.5-flash"

URL = f"https://generativelanguage.googleapis.com/v1beta/models/{MODEL}:generateContent?key={API_KEY}"

# A simple chat prompt to test token usage
payload = {
    "contents": [
        {"role": "user", "parts": [{"text": "You are a helpful mental health assistant. Hello, how are you?"}]}
    ],
    "generationConfig": {
        "maxOutputTokens": 512,
        "temperature": 0.4
    }
}
headers = {"Content-Type": "application/json"}

print("Sending request to Gemini API...")
response = requests.post(URL, data=json.dumps(payload), headers=headers)

if response.status_code == 200:
    res_json = response.json()
    
    print("\n=== SUCCESS ===")
    print("\n--- Token Usage ---")
    print(json.dumps(res_json.get('usageMetadata', {}), indent=2))
    
    print("\n--- Model Response ---")
    print(res_json['candidates'][0]['content']['parts'][0]['text'])
else:
    print(f"\n=== ERROR {response.status_code} ===")
    print(response.text)
```
